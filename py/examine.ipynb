Depression som utgångspunkt

- Hur stor andel av studenterna är deprimerade?
- Är depression vanligare bland män eller kvinnor?
- Skiljer sig medelåldern mellan deprimerade och icke-deprimerade?

```python
# Hur stor andel av studenterna är deprimerade?
with sqlite3.connect(db_path) as conn:
    result = conn.execute("SELECT COUNT(*) FROM students WHERE depression = 1").fetchone()[0]
    total_students = conn.execute("SELECT COUNT(*) FROM students").fetchone()[0]
    depression_percentage = (result / total_students) * 100
    print(f"Andel deprimerade studenter: {depression_percentage:.2f}%")
```

```python
# Är depression vanligare bland män eller kvinnor?
with sqlite3.connect(db_path) as conn:
    male_depression = conn.execute("SELECT COUNT(*) FROM students WHERE depression = 1 AND gender = 'Male'").fetchone()[0]
    female_depression = conn.execute("SELECT COUNT(*) FROM students WHERE depression = 1 AND gender = 'Female'").fetchone()[0]
    total_male = conn.execute("SELECT COUNT(*) FROM students WHERE gender = 'Male'").fetchone()[0]
    total_female = conn.execute("SELECT COUNT(*) FROM students WHERE gender = 'Female'").fetchone()[0]

    male_percentage = (male_depression / total_male) * 100 if total_male > 0 else 0
    female_percentage = (female_depression / total_female) * 100 if total_female > 0 else 0

    print(f"Andel deprimerade män: {male_percentage:.2f}%")
    print(f"Andel deprimerade kvinnor: {female_percentage:.2f}%")
```

```python
# Skiljer sig medelåldern mellan deprimerade och icke-deprimerade?
with sqlite3.connect(db_path) as conn:
    depressed_age = conn.execute("SELECT AVG(age) FROM students WHERE depression = 1").fetchone()[0]
    non_depressed_age = conn.execute("SELECT AVG(age) FROM students WHERE depression = 0").fetchone()[0]

    print(f"Medelålder för deprimerade studenter: {depressed_age:.2f} år")
    print(f"Medelålder för icke-deprimerade studenter: {non_depressed_age:.2f} år")
```
